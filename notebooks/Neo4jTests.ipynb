# 2. Create a test graph

```python
from neomodel import (
    StructuredNode, 
    StructuredRel, 
    StringProperty, 
    FloatProperty, 
    RelationshipTo, 
    get_config
)
from neo4j import basic_auth

config = get_config()
connection_string = 'bolt://' + CREDENTIALS["NEO4J_USER"] + ':' + CREDENTIALS["NEO4J_PASS"] + '@neo4j:7687'
config.database_url = connection_string
```

```python
# Rest of your code remains the same
class InterestRel(StructuredRel):
    weight = FloatProperty(default=1.0)

class Book(StructuredNode):
    title = StringProperty(unique_index=True, required=True)

class Person(StructuredNode):
    name = StringProperty(index=True)
    likes = RelationshipTo(Book, 'LIKES', model=InterestRel)

matrix = Book(title='The Matrix Explored').save()
bob = Person(name='Bob').save()
bob.likes.connect(matrix, {'weight': 0.85})
```
