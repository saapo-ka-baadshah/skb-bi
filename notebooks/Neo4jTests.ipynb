# 2. Create a test graph

```python
from neomodel import (
    StructuredNode, 
    StructuredRel, 
    StringProperty, 
    FloatProperty, 
    RelationshipTo, 
    get_config
)

config = get_config()

config.DATABASE_URL = f"bolt://{CREDENTIALS['NEO4J_USER']}:{CREDENTIALS['NEO4J_PASS']}@{CREDENTIALS['NEO4J_HOST']}:7687"

# 1. Define the relationship properties
class InterestRel(StructuredRel):
    weight = FloatProperty(default=1.0)  # Your edge weight property

# 2. Reference the custom relationship in your Node model
class Book(StructuredNode):
    title = StringProperty(unique_index=True, required=True)

class Person(StructuredNode):
    name = StringProperty(index=True)
    
    # Pass your custom relationship class to the model type
    likes = RelationshipTo(Book, 'LIKES', model=InterestRel)

# 3. Create nodes and apply a weighted relationship
matrix = Book(title='The Matrix Explored').save()
bob = Person(name='Bob').save()

# Pass properties directly into the connect() method
bob.likes.connect(matrix, {'weight': 0.85}) 
```
